```python
import numpy 
ids=numpy.array([101,102,103,104,105])
names=numpy.array(['Aysel','Murad','Leyla','Tural','Nigar'])
scores=numpy.array([78,92,65,88,71])
```

```python
#2
for i in range (len(ids)):
    print (f' name: {names[i]} ids: {ids[i]} scores: {scores[i]}')
```

```python
#7
highest=numpy.max(scores)
lowest=numpy.min(scores)
average=round (numpy.mean(scores),2)
print(f'highest score:{highest} lowest score:{lowest} average score:{average}')
```

```python
#8
scores=numpy.sort(scores)
print(scores[::-1])
```

```python
#9
passing_ones=[]
pass_count=0
for i in scores:
    if i>=70:
        passing_ones+=[i]
        pass_count+=1
print(passing_ones,' ',pass_count)
passing_ones=passing_ones.tolist()
print(passing_ones)
```
